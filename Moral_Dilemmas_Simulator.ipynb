```python
# Step 1: Import libraries
import json  # For saving results in JSON format
import csv   # For saving results in CSV format
from datetime import datetime  # To timestamp responses
```

```python
# Step 2: Define moral dilemmas
moral_dilemmas = [
    {
        "id": 1,
        "scenario": "A runaway trolley is heading towards 5 people. You can pull a lever to divert it onto another track where 1 person is tied. What do you do?",
        "options": ["Do nothing (5 die)", "Pull the lever (1 dies)"]
    },
    {
        "id": 2,
        "scenario": "You see a wallet on the street with 100€. Do you return it or keep it?",
        "options": ["Return it", "Keep it"]
    },
    {
        "id": 3,
        "scenario": "You are a doctor with 5 patients needing different organs. A healthy person comes in for a check-up. Do you sacrifice them to save 5?",
        "options": ["Do not sacrifice", "Sacrifice to save 5"]
    }
]
```

```python
# Step 3: Function to present dilemmas and collect answers
def present_dilemmas(dilemmas):
    responses = []  # To store user responses

    for dilemma in dilemmas:
        print("\nScenario:", dilemma["scenario"])
        for i, option in enumerate(dilemma["options"], start=1):
            print(f"{i}. {option}")

        # Input validation loop
        while True:
            choice = input("Enter the number of your choice: ")
            if choice.isdigit() and 1 <= int(choice) <= len(dilemma["options"]):
                choice = int(choice)
                break
            else:
                print("Invalid input, please enter a valid number.")

        # Save response with timestamp
        responses.append({
            "dilemma_id": dilemma["id"],
            "choice": dilemma["options"][choice-1],
            "timestamp": str(datetime.now())
        })

    return responses
```

```python
# Step 4: Import necessary libraries for saving
import csv
import json
from datetime import datetime
```

```python
# Step 5: Function to save responses in CSV
def save_responses_csv(responses, filename="moral_dilemmas_responses.csv"):
    # Define CSV headers
    headers = ["dilemma_id", "choice", "timestamp"]

    # Open CSV file in write mode
    with open(filename, mode='w', newline='') as file:
        writer = csv.DictWriter(file, fieldnames=headers)
        writer.writeheader()  # Write headers
        writer.writerows(responses)  # Write all responses

    print(f"Responses saved to CSV: {filename}")
```

```python
# Step 6: Function to save responses in JSON
def save_responses_json(responses, filename="moral_dilemmas_responses.json"):
    with open(filename, 'w') as file:
        json.dump(responses, file, indent=4)  # Save with indentation for readability
    print(f"Responses saved to JSON: {filename}")
```

```python
# Step 7: Run the moral dilemma simulator
user_responses = present_dilemmas(moral_dilemmas)  # Present dilemmas
save_responses_csv(user_responses)  # Save CSV
save_responses_json(user_responses)  # Save JSON
```
